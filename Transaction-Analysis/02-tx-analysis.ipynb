```python
# ¿Cuál es el monto total y el monto promedio de las transacciones fraudulentas (is_fraud=1) comparado con las no fraudulentas?

query = """
SELECT
    fl.is_fraud,
    SUM(tx.amount) AS total,
    AVG(tx.amount) AS avg
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
GROUP BY
    fl.is_fraud
"""
rslt = spark.sql(query).show()
```

```python
# ¿Cuántas transacciones tiene cada card_id? ¿Cuál tarjeta es la más usada?

query = """
SELECT
    cd.card_type,
    COUNT(tx.tx_id)
FROM transactions AS tx
LEFT JOIN cards AS cd
    ON cd.card_id = tx.card_id
GROUP BY
    cd.card_type
"""
rslt = spark.sql(query).show()
```

```python
# ¿Cuántos usuarios tienen más de una tarjeta asociada en la tabla cards?

query = """
SELECT
    usr.full_name,
    cd.card_type,
    COUNT(cd.card_id)
FROM cards AS cd
LEFT JOIN users AS usr
    ON usr.user_id = cd.user_id
GROUP BY
    usr.full_name,
    cd.card_type
ORDER BY
    usr.full_name
"""
rslt = spark.sql(query).show()
```

```python
# Lista las transacciones cuyo país de la transacción (transaction_details.country) es distinto al país registrado del usuario (users.country).

query =  """
SELECT *
FROM transactions AS tx
LEFT JOIN transaction_details AS td
    ON td.tx_id = tx.tx_id
LEFT JOIN users AS usr
    ON usr.user_id = tx.user_id
WHERE
    td.country != usr.country
"""
rslt = spark.sql(query).show()
```

```python
# ¿Qué porcentaje del monto total transaccionado corresponde a transacciones marcadas como fraude?

query = """
SELECT
    SUM(tx.amount)
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
"""
rslt = spark.sql(query).show()
```

```python
query = """
SELECT
    fl.is_fraud AS is_fraud,
    SUM(tx.amount) AS total,
    (
    SELECT
        SUM(tx.amount)
    FROM transactions AS tx
    ) AS total_tx
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
GROUP BY
    fl.is_fraud
"""
rslt = spark.sql(query).show()
```

```python
query = """
WITH table AS (
    SELECT
        fl.is_fraud AS is_fraud,
        SUM(tx.amount) AS total,
        (SELECT
            SUM(tx.amount)
        FROM transactions AS tx) AS total_tx
    FROM transactions AS tx
    LEFT JOIN fraud_labels AS fl
        ON fl.tx_id = tx.tx_id
    GROUP BY
        fl.is_fraud
)

SELECT
    *,
    ROUND((t.total * 100)/t.total_tx,2)
FROM table as t

"""
rslt = spark.sql(query).show()
```

```python
# ¿Cuántas transacciones DECLINED tiene cada usuario? ¿Quién tiene más?

query = """
SELECT *
FROM transactions AS tx
LEFT JOIN users AS usr
    ON usr.user_id = tx.user_id
"""
rslt = spark.sql(query).show()
```

```python
query = """
SELECT
    usr.full_name AS name,
    COUNT(CASE WHEN tx.status::string == 'DECLINED' THEN 1 END) AS status
FROM transactions AS tx
LEFT JOIN users AS usr
    ON usr.user_id = tx.user_id
GROUP BY
    name
ORDER BY
    status DESC
"""
rslt = spark.sql(query).show()
```

```python
# ¿Cuál es la fecha de la primera transacción fraudulenta registrada en todo el dataset?

query =  """
SELECT *
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
WHERE
    fl.is_fraud == 1
ORDER BY
    tx.tx_date ASC
"""
rslt = spark.sql(query).show()
```

```python
# Calcula la tasa de fraude (%) agrupando por card_type (crédito vs débito).

query = """
SELECT *
FROM transactions AS tx
LEFT JOIN cards AS cd
    ON cd.card_id = tx.card_id
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
"""
rslt = spark.sql(query).show()
```

```python
query = """
SELECT
    cd.card_type AS card_type,
    COUNT(CASE WHEN fl.is_fraud == 1 THEN 1 END) AS count_isfrd,
    COUNT(*) AS total_tx
FROM transactions AS tx
LEFT JOIN cards AS cd
    ON cd.card_id = tx.card_id
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
GROUP BY
    cd.card_type
"""
rslt = spark.sql(query).show()
```

```python
query = """
WITH table AS (
    SELECT
        cd.card_type AS card_type,
        COUNT(CASE WHEN fl.is_fraud == 1 THEN 1 END) AS count_isfrd,
        COUNT(*) AS total_tx
    FROM transactions AS tx
    LEFT JOIN cards AS cd
        ON cd.card_id = tx.card_id
    LEFT JOIN fraud_labels AS fl
        ON fl.tx_id = tx.tx_id
    GROUP BY
        cd.card_type
)

SELECT
    *,
    ROUND(((t.count_isfrd * 100)/t.total_tx),2) AS fraud_rate
FROM table AS t
"""
rslt = spark.sql(query).show()
```

```python
# Calcula la tasa de fraude (%) agrupando por category del merchant (Retail, E-commerce, Streaming, Gaming, etc.).

query = """
SELECT
    mc.category AS category,
    COUNT(tx.tx_id) AS tx,
    COUNT(CASE WHEN fl.is_fraud == 1 THEN 1 END) AS fraud_tx
FROM transactions AS tx
LEFT JOIN merchants as mc
    ON mc.merchant_id = tx.merchant_id
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
GROUP BY
    category
"""
rslt = spark.sql(query).show()
```

```python
# 10. Extrae la hora del día (0–23) de tx_date y calcula cuántas transacciones totales y cuántas fraudulentas hay por hora. ¿En qué franja se concentra más el fraude?

query = """
SELECT
    EXTRACT(hour FROM tx.tx_date) AS hour,
    COUNT(tx.tx_id) AS tx,
    COUNT(CASE WHEN fl.is_fraud == 1 THEN 1 END) AS fraud_tx
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
GROUP BY
    hour
ORDER BY
    hour DESC
"""
rslt = spark.sql(query).show()
```

```python
# 11. Encuentra usuarios que, en un mismo día, hicieron transacciones con 3 o más merchants distintos.

query = """
SELECT
    tx.user_id AS user,
    CAST(tx.tx_date AS DATE) AS date,
    COUNT(DISTINCT(tx.merchant_id)) AS count
FROM transactions AS tx
GROUP BY
    tx.user_id,
    CAST(tx.tx_date AS DATE)
HAVING
    count >= 3
"""
rslt = spark.sql(query).show()
```

```python
# 12. Identifica combinaciones de (user_id, merchant_id, amount) que se repiten más de una vez — posibles cargos duplicados.

query = """
SELECT
    STRUCT(tx.user_id, tx.merchant_id, tx.amount),
    COUNT(*)
FROM transactions AS tx
GROUP BY
    STRUCT(tx.user_id, tx.merchant_id, tx.amount)
HAVING
    COUNT(*) > 2
"""
rslt = spark.sql(query).show()
```

```python
# 13. Compara el monto promedio de transacciones APPROVED vs DECLINED, por usuario.

query = """
SELECT
    tx.user_id,
    tx.status,
    ROUND(AVG(tx.amount),2)
FROM transactions AS tx
GROUP BY
    tx.user_id,
    tx.status
"""
rslt = spark.sql(query).show()
```

```python
# 14. Cruzando fraud_labels con users, calcula el monto total perdido por fraude (solo status='APPROVED' e is_fraud=1) agrupado por risk_level.

query = """
SELECT
    usr.risk_level,
    SUM(tx.amount)
FROM transactions AS tx
LEFT JOIN fraud_labels AS fl
    ON fl.tx_id = tx.tx_id
LEFT JOIN users AS usr
    ON usr.user_id = tx.user_id
WHERE
    tx.status = 'APPROVED'
    AND fl.is_fraud = 1
GROUP BY
    usr.risk_level
"""
rslt = spark.sql(query).show()
```
