```python
def solution(N):
    N = bin(N)[2:]
    print(N)

    gp = 0
    maxgp = 0

    for k in N:
        if int(k) == 0:
            gp += 1
        elif int(k) == 1:
            maxgp = max(gp, maxgp)
            gp = 0
    
    return maxgp


solution(226)
```

## Cycling array rotation

```python
def solution(A, k):
    n = len(A)
    print(A)

    B = [None]*n

    for i in range(0, n):
        print((i+k)%n)
        B[(i+k)%n] = A[i]

    return B

solution([1,2,3,4,5], 2)
```

## Odd Ocurrence array in python

```python
def solution(A):
    if len(A) == 1:
        return A[0]

    A.sort()
    for i in range(0, len(A)-1, 2):
        if A[i] != A[i+1]:
            return A[i]

    return A[-1]


solution([3,4,3,5,5])
```

## Frog jump

```python
def solution(x,y,d):
    v = (y-x)//d

    if x + (v*d) >= y:
        return print(f'# jumps {v}')
    else:
        return print(f'# jumps {v} + one additional')


solution(2,9,2)
```

## Perm missing element

```python
def solution(A):
    if (len(A) == 0):
        return 1

    A.sort()
    for i in range(0, len(A)):
        if A[i] != i+1:
            return i+1

    return len(A)+1

solution([1,2,3,4]) 
```
